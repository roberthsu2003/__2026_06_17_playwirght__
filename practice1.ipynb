```python
import twstock

# 查詢特定股票代碼資訊
stock_info = twstock.codes['2330']
print(stock_info.name)       # 台積電
print(stock_info.market)     # 上市
print(stock_info.group)      # 半導體事業

# 搜尋產業或關鍵字
for code, info in twstock.codes.items():
    if '半導體' in info.group:
        print(f"{code}: {info.name}")
```

```python
import yfinance as yf

# 1. 設定股票代碼（以台積電 2330 為例，上市股票需加上 .TW）
stock_id = "2330.TW"

# 2. 建立 Ticker 物件
stock = yf.Ticker(stock_id)

# ----------------------------------------------------
# 方法 A：使用相對時間區間下載（例如：過去一個月 "1mo"、過去一年 "1y"）
# ----------------------------------------------------
df_1mo = stock.history(period="1mo")

print("=== 過去一個月歷史價格 ===")
# 顯示前 5 筆資料（包含 Date, Open, High, Low, Close, Volume 等欄位）
print(df_1mo.head())

# ----------------------------------------------------
# 方法 B：指定明確的開始與結束日期（格式：YYYY-MM-DD）
# ----------------------------------------------------
df_custom = yf.download(stock_id, start="2026-01-01", end="2026-08-01")

print("\n=== 指定日期區間歷史價格 ===")
print(df_custom.head())
```
